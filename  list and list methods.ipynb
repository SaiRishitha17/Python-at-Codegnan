```python
# Define List
# List is ordered collection
# List allows indexing and slicing
lst = [1,2,4.5,"Hi","Hello",1,3,4]
print(lst)
print(lst[2]) # indexing
print(lst[1:5:1]) #slicing
print(lst[-2:6])
```

```python
# slicing
lst = [10,20,30,40,50]
print(lst[:5:-1])
print(lst[::-1])
print(lst[:500])
```

```python
#empty list creation
l1 = []
l2 = list()
print(type(l1), type(l2))
```

```python
# List operations
#1. List concatination (using + operator)
#2. List Repetition (using * operator)
l1 = [1,2,3]
l2 = [3,4,5,6]
print(l1+l2)
print(l1 * 2)
print([0] * 10)
```

### List Methods
- Item adding methods
    - append(item) : It adds item at end of list
    - extend(iterator): It adds all the iterator elements at the end of the list i.e multiple items
    - insert(pos, item): It adds item at specified position #when this method is performed,in the mentioned intex the number we need is replaced and rest of the elements take a right shift
      
  

```python
lst = [10,20,30,40]
lst.append(100)
print(lst)
lst.extend([1,2,3])
print(lst)
lst.append([100,200])
print(lst)
lst.insert(0,1000)
print(lst)
```

- Item Removing Methods
    - remove(item): It removes the first occurance Item from the list if it is existing in the list otherwise "valueError"
    - pop(pos):It removes and return the specified position item if position is passed,otherwise it remove and returns the last item from list
    - clear(): All elements from list
  

```python
lst = [10,20,30,40]
print(lst.remove(20)) #because it is performing some operation,the output is none asit internally performed the operation
print(lst)      #here we are printing the output after the operation so we get the updated list
```

```python
# pop
lst = [20,30,40,50,20]
val = lst.pop()
print(val)
print(lst)
val2 = lst.pop(0) #It removes the element in 0th position
print(val2)
print(lst)
lst.clear()
print(lst)
```

```python
# index and count()
# index(item): If item is present in list it returns the index otherwise value error
#count(): It returns the count of Total occurances of item
lst = [10,20,30,20,40,50]
ind = lst.index(20)
print(lst.count(20))
print(ind)
```

```python
# Sorting items
# sort(reverse = False) :By default it sorts the original list into Ascending order
#                             if reverse is True it sorts in descending order
lst = [1,4,2.5,-5,3,-45,-4,30]
lst.sort()
print(lst)
lst.sort(reverse = True)
print(lst)
```

```python
# Sorting items # if list contains strings
# sort(reverse = False) :By default it sorts the original list into Ascending order
#                             if reverse is True it sorts in descending order
lst = ["hi","hello","bunty"]
lst.sort()
print(lst)
lst.sort(reverse = True)
print(lst)
```

```python
# Builtin Functions- Len(), min(), max(), sum() # because these are builtin fun lst is given inside paranthesis
lst = [1,2,3,4,5]
print(len(lst),min(lst), max(lst), sum(lst))
print("Average:",sum(lst) / len(lst))
```
