- Methods: The Function defination is implemented inside the class those functions called as methods.
    - Types of Methods
      1. Instance Method
      2. class Method
      3. static Method


```python
# class attribute - school
# instance attributes - id,name,class
class Student:
    school = "Slate The school"
    def __init__(self,name,roll,marks):
        self.name = name
        self.roll = roll
        self.marks = marks
    def student_details(self):
        return f"My Self {self.name}  bearing roll no- {self.roll} scored {self.marks} marks "
s = Student("Rishitha",735047,100)
print(s.name)
print(s.roll)
print(s.marks)
print(s.student_details())
        
        
```

```python
# class attribute - school
# instance attributes - id,name,class
# Instance Method
# If the method takes "self" inside ,then it comes under instance method
class Student:
    school = "Slate The school" #class attribute
    def __init__(self,name,roll,marks):
        self.name = name #instance attribute
        self.roll = roll #instance attribute
        self.marks = marks #instance attribute
    def student_details(self): #instance method
        return f"My Self {self.name}  bearing roll no- {self.roll} scored {self.marks} marks "
#class Method
    @classmethod
    def update_school(cls,new_school_name):
        cls.school = new_school_name
        return "School name Updated"
#static Method
    @staticmethod
    def percentage(marks:list):
        return (sum(marks)/300)*100

s1 = Student("Rishitha",735047,80)
s2 = Student("priya" ,735048,85)
print(s1.school, s2.school)
print(s.student_details())
print(s1.update_school("SRDIGI school"))
print(s1.school, s2.school)
print(s2.percentage ([80,85,100]))
```

- `Instance Method`: The Method defination takes first parameter as `self` those methods are called as `Instance Method`.This method works on instance attributes
  
   `syntax:` <br>
     `         class class_name:`           <br>
  `              def method_name(self):`   <br>
  `                      #block code`
- `Class Method`: The method defination starts with class method decorator and it takes first parmeter as "cls", those methods called as "Class Methods" .This methods works on class attributes
  
   `syntax:` <br>
  `        class class_name:`            <br>
  `             @classmethod`      <br>
  `            def method_name(cls,parameters):`    <br>
  `              #block code`
- `Static Method`:The method defination starts with `staticmethod` decorator and it doesnt take first parameter as self or cls those methods are called as static methods.This methods work on except instance attributes and class attributes.
  
   ` syntax:` <br>
  `        class class_name:`            <br>
  `             @staticsmethod`      <br>
  `            def method_name(cls,parameters):`    <br>
  `              #block code`
- `Initialisation constructor(__init__)`: the initialisation constructor is used to initialise the instance attributes data and this constructor takes first parameter as `self` and this constructor executes whenever an object is created for the class.


- ### Encapsulation:
    - Encapsulation is wrapping the methods and attributes in single unit that is class.
    - `The main agenda of Encapsulation is to restric the direct access of attributes and methods`
    - This type of restriction can be implemented by using `Access Modifiers`
  -  `Types of Access Modifiers`
        1. Public()
        2. Protected()  #starts with single underscore
        3. Private()    #starts with dunderscore
    - In Python we dont have any keywords to represent it as public, protected and private but by using underscore(_) we can define
      1. `Public` : The Public Members(i.e attributes + Methods) does not starts with underscore(_). The public members can be accessed anywhere through out the program.(i.e inside or outside the class we can access)
      2. `Protected` : The protected Members(i.e attributes + Methods) starts with single underscore.This members can access inside the class and in subclass also
      3. `Private` : The private Members(i.e attributes + Methods) starts with double underscore(__).This members can accessonly inside the class
- `Accessing levels of Access Modifiers`:
  - public: inside class, sub class, outside the class
  - protected: inside class, sub class
  - private: only inside the class
  

```python
class Bank:
    def __init__(self, name,account,password,balance):
        self.name = name
        self.account = account
        self.__password = password
        self._balance = balance
    def __total_users(self):  #getter method
        return "total users : 101"
    def get_balance(self):  #getter method
        return f'Current Balance is:{self._balance}'
    def deposite(self, deposite_amount):   #setter method
        self._balance += deposite_amount
        return f"{deposite_amount} is: Deposited Successfully"

u1 = Bank("Rishi", 2546, "rishi@455", 10000)
print(u1.name)
print(u1.get_balance())
print(u1.deposite(5000))
print(u1.get_balance())
print(u1._balance) #protected members access in outside the class
print(u1._Bank__total_users()) #Name Mangling :syntax: obj._classname__attribute)
```

```python
#ATM pin system
class ATM:
    def __init__(self, pin):
        self.__pin = pin
    def verify_pin(self, entered_pin):
        if entered_pin == self.__pin:
            return "Access Granted"
        return "Incorrect PIN"
    def change_pin(self, old_pin, new_pin):
        if old_pin == self.__pin:
            self.__pin = new_pin
            return "PIN Changed Successfully"
        return "Wrong Old PIN"
atm = ATM(1234)

print(atm.verify_pin(1234))
print(atm.change_pin(1234, 5678))
print(atm.verify_pin(5678))
```

- Python does not support high security like Java.Here we can access protected memmbers and private members outside the class by usinf following ways
- To access the protected member   syntax - `obj._protected member name`
- To access the private members by using name mangling   syntax - `obj._classname__privatemember name`
- `Advantages:`
    - provide security for data and functionalities
