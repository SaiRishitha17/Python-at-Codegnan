```python
#O
import re

# Read the input line (e.g., "7+54" or "100/5")
line = input().strip()

# Split the string into numbers and the operator
# This looks for numbers (\d+) and non-digit characters ([^0-9])
parts = re.split(r'(\+|-|\*|/)', line)

# parts[0] is A, parts[1] is the operator S, parts[2] is B
a = int(parts[0])
s = parts[1]
b = int(parts[2])

# Perform the operation based on the operator S
if s == '+':
    print(a + b)
elif s == '-':
    print(a - b)
elif s == '*':
    print(a * b)
elif s == '/':
    # Use // for integer division to satisfy the "without fractions" rule
    print(a // b)

```
