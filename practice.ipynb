```python
credit_score = int(input("Enter your credit score"))
income = int(input("Enter your income"))
liabilities = int(input("Enter your liabilities "))
if credit_score >= 750 and income>=50000 and liabilities<=20000:
    print("Approved")
elif 650<=credit_score<750 and income>=50000 and liabilities<=20000:
    print("approved with conditions")
else:
    print('reject')
    


```
