```python
##### who will reach top(100) in less steps
import random
count = 0
rolls = 0
print("welcome to the who will reach top in less steps Game")
while True:
    input('To throw a dice,please click enter button')
    face_value=random.randint(1,6)
    
    if count+face_value <= 100:
        count += face_value
    rolls+=1
    if count == 100:
        print(f"You have reached the top in {rolls} chances")
        break
    print(f'Your dice value:{face_value}')
    print(f"You are at {count} position in {rolls} chances")
    print('*'* 50)
```

```python
##snake and ladder
import random
position = 0
rolls = 0
print("SNAKE AND LADDER")
ladder = {8:15,12:22,35:62}
snake = {34:22,55:34,78:47}
while True:
    input('To throw a dice,please click enter button')
    face_value=random.randint(1,6)
    if position+face_value <= 100:
        position += face_value
    rolls+=1
    if position in ladder:
        print(f" Ladder! You climbed from {position} to {ladder[position]}")
        position = ladder[position]
    if position in snake:
        print(f"snake! You slid from {position} to {snake[position]}")
        position = snake[position]
    if position == 100:
        print(f"You have reached the target in {rolls} chances")
        break
    print(f"Your dice value: {face_value}")
    print(f"You are at position {position} in {rolls} chances")
    print("*" * 50)

    

```
