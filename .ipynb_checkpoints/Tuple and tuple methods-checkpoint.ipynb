```python
# charactersitics
# Ordered collection
t = (1,2,3,[10,30,40],23)
t[3]
```

```python
# immutable
# after operation it doesnot change the lenght of object and doesnt change remaining element position
t = (1,2,3,[10,30,40],23)
t[3].append(100)
print(t)
```

```python
#empty tuple representation
#single element representation
t1 = ()
t2 = tuple()
print(type(t1),type(t2))
```

```python
#single element representation
t1 = (10)
t2 = (10,) #to represent tuple we need to use , beside else it will read it as integer
l1 = [10]
print(type(t1),type(t2),type(l1))
```

```python
(0,) * 10  # tuple 
```

```python
#Tuple operations
#tuple concatination
#tuple repetation
t1 = (2,3.4,25,[12,20,30],"hi")
t2 = (0,45,7,-3,"hello")
t3 = (t1+t2)
print(t3)
t1 = (1,2,3)
t2 = (1,2,3)
t3 = (t1+t2)
print(t3)


#repetation
t1 = (1,2,3,-7)
print(t1*10)
```

```python
#tuple methods
#index()
#count()
t1 = (11,22,33,"hi","hello")
ind = t1.index(22)
print(ind)

t2 = (10,20,0,0,0,0,0,34,0,0,0,-1)
count = t2.count(0)
print(count)
```

```python
# Builtin Functions- Len(), min(), max(), sum() # because these are builtin fun lst is given inside paranthesis
t1 = (1.2,3,4,7,-4,25,78,42)
print(len(t1),min(t1), max(t1), sum(t1))
print("average.", sum(t1) / len(t1))

```
