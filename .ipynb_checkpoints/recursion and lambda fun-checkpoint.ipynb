- ### Direct recursion

```python
#print 1 to n using recursion
def func(x):
    if x == 0:
        return 0
    func(x-1)
    print(x)
func(3)
```

```python
# factorial
def factorial(x):
    if x==0 or x==1:
        return 1
    return x*factorial(x-1)
print(factorial(5))
```

```python
#sum of digits in a number
# n = 456 o/p:5
def sumOfDigits(n):
    if n == 0:
        return 0
    return n%10 + sumOfDigits(n//10)
print(sumOfDigits(456))
print(sumOfDigits(123456))

```

```python
#count digits in a number
# n = 456 o/p:5
def sumOfDigits(n):
    if n == 0:
        return 0
    return 1 + sumOfDigits(n//10)
print(sumOfDigits(456))
print(sumOfDigits(123456))

```

- ### Lambda Functions or Anynomous Function

```python
# find the square of a number using lambda
s = lambda  x:x**2
print(s(10))
```

```python
arr = [1,2,3,4,5]
print(list(map(lambda x:x**2, arr)))
```

```python
arr = [1,2,3,4,5]
print(list(filter(lambda x:x%2==0, arr)))
```

```python
#check weather the number is even or odd using lamda function
s = lambda x:"even" if x%2==0 else"odd"   #used oneline if-else statement
print(s(10))
```

```python

#1 - odd
#2 - even
arr = [1,2,3,4,5]
print(list(map(lambda x:f"{x} -even" if x%2==0 else f"{x} - odd" ,arr)))
```

```python
string = input()
count = {}
for char in string:
    if char in count:
        count[char] += 1
    else:
        count[char] = 1
print(count)
```

```python
string = input()
count = {}
for char in string:
        count[char] = count.get(char,0)+1
print(count)
```
