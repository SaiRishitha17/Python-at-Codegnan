`Exception`: Exception is nothing but it is an error,whenever an exception occurs it stops the program execution and it returns type of exception and followed by execution message.

- `Types of Errors:`
  1. syntax error or compiletime error
  2. Runtime error
  3. Logical error

`1.Syntax Errors:` The syntax error raises during the compilation time of the program,those errors we can call it as syntax errors or compiletime errors

   eg:Intendation error, syntax error
   
`2.Runtime Errors/Runtime Exception:`  The runtime error raises during the runtime of a program,those errors are called as runtime errors
 - we cannot handle syntax errors but we can handle the runtime errors by using Exception handling blocks (i.e.,try and except block)

   eg:keyError, valueError, indexError, attributeError, zerodivisonError... etc
   
`EXCEPTION HANDLING BLOCKS:`

        syntax:
           try:    
                    #block of code that might contain exception   
           except:    
                    #exception Handling block         
           else:   (optional)
                    #this blocks exectues when there is no exception raised in try block  
           finally: (optional)  
                    #this block always executes every exception raised or not
    
   

`3.Logical Errors`:
