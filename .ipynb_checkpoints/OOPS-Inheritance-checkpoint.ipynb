- ## Inheritance:

- In python,a class get properties from another class(i.e.,from parent class to child class)
  
`Types of Inheritance`
 - 1. Single Inheritance
   2. Multi-level Inheritance
   3. Multiple Inheritance
   4. Hierarchial Inheritance
   5. Hybrid Inheritance

- `1.Single Inheritance`: A child class get properties from single parent class.
- 
  `syntax:`<br>
                 `class parent_class_name:` <br>
                            # attributes           <br>
                            # methods              <br>
                 `class childclass_name(parent_class_name)`   <br>
                             # attributes                            <br>
                             # methods
- `2.Multilevel Inheritance`: A class (i.e.,child class) get properties from another class (i.e.,parent class) that class get properties from some another class(i.e.,Grand parent)
- `3.Multiple Inheritance` :A single chile class get properties from multiple parent classes which is called as multiple
inheritance
- `4.Hierarchial Inheritance`:single parent class and multiple child classes 
- `5.Hybrid Inheritance`:The combination of two or more types of inheritance 
  

```python
#single inheritance
#without child class implementation
#when child class dont have constructor ,in that case parent class constructor acts as a child class constructor
class Person:   #parent class
    def __init__(self,name,age,gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):  #child classs
        return f'My name is {self.name} and I am {self.age} years old'
        
class Student(Person):
    pass
s = Student("rishi",21,"Female")
print(s.name)
print(s.details())
```

```python
#single inheritance
#with child class implementation
#child class with constructor: acts individually because we have initialised init for both so both acts individually
class Person:   #parent class
    def __init__(self,name,age,gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):  #child classs
        return f'My name is {self.name} and I am {self.age} years old'
        
class Student(Person):
    def __init__(self,stdid,name,age,branch):
        self.name = name
        self.age = age
        self.stdid = stdid
        #self.gender = gender
        self.branch = branch 
s = Student(101,"rishi",21,"ECE")
print(s.name)
print(s.branch)
print(s.age)
print(s.stdid)
#print(s.gender)   AttributeError: 'Student' object has no attribute 'gender'
print(s.details())
```

```python
#single inheritance
#with child class implementation
#child class with constructor: acts individually because we have initialised init for both so both acts individually
class Person:   #parent class
    def __init__(self,name,age,gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):  #child classs
        return f'My name is {self.name} and I am {self.age} years old and i am {self.gender}'
        
class Student(Person):
    def __init__(self,stdid,name,age,branch):
        self.name = name
        self.age = age
        self.stdid = stdid
        #self.gender = gender
        self.branch = branch 
s = Student(101,"rishi",21,"ECE")
print(s.name)
print(s.branch)
print(s.age)
print(s.stdid)
print(s.gender)   #AttributeError: 'Student' object has no attribute 'gender'

```

```python
# Single Inheritance

class Employee:   # Parent Class
    def __init__(self, name, emp_id):
        self.name = name
        self.emp_id = emp_id
    def employee_details(self):
        return f"Employee Name: {self.name}\nEmployee ID: {self.emp_id}"
class Developer(Employee):   # Child Class
    def __init__(self, name, emp_id, programming_language):
        super().__init__(name, emp_id)
        self.programming_language = programming_language

    def developer_details(self):
        return (f"{self.employee_details()}\n"
            f"Programming Language: {self.programming_language}")
# Object Creation
developer = Developer("Rishitha", 101, "Python")

print(developer.developer_details())
```

```python
# multilevel inheritance
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def details(self):
        return f"Myself {self.name} and i am {self.age} years old"
class Student(Person):
    def __init__(self,name,age,branch):
        super().__init__(name,age)
        self.branch = branch
class Employee(Student):
    def __init__(self,name,age,branch,salary):
        super().__init__(name,age,branch)
        self.salary = salary
    def get_salary(self, Bonous=0):
        self.salary += Bonous
        return self.salary
e = Employee("rishi",20,"hyd",50000)
print(e.name)
print(e.branch)
print(e.details())
print(e.get_salary())
print(e.get_salary(Bonous = 10000))
```

```python
#Multiple inheritance
class Father:
    def __init__(self, Father_name):
        self.Father_name = Father_name
class Mother:
    def __init__(self, Mother_name):
        self.Mother_name = Mother_name
#child class
class son(Father, Mother):
    def __init__(self,name,age,Father_name, Mother_name):
        Father.__init__(self,Father_name)
        Mother.__init__(self,Mother_name)
        self.name=name
        self.age=age
s = son("siva", 32,"pavan","satya")
print(s.Father_name)
print(s.Mother_name)

    
```

```python
#Multiple inheritance
class Father:
    def __init__(self, Father_name):
        self.Father_name = Father_name
    def work(self):
        return "Father is an Engineer"
class Mother:
    def __init__(self, Mother_name):
        self.Mother_name = Mother_name
    def work(self):
        return "Mother is a Teacher"
#child class
class son(Father, Mother):
    def __init__(self,name,age,Father_name, Mother_name):
        Father.__init__(self,Father_name)
        Mother.__init__(self,Mother_name)
        self.name=name
        self.age=age
    def parent_work(self):
        print(Father(self.Father_name).work())
        print(Mother(self.Mother_name).work())
s = son("siva", 32,"pavan","satya")
print(s.Father_name)
print(s.Mother_name)
print(s.work())
print(s.work())
print(s.parent_work())
print(son.mro())
```

```python
#Multiple inheritance
class Father:
    def __init__(self, Father_name):
        self.Father_name = Father_name
    def work1(self):
        return "Father is an Engineer"
class Mother:
    def __init__(self, Mother_name):
        self.Mother_name = Mother_name
    def work2(self):
        return "Mother is a Teacher"
#child class
class son(Father, Mother):
    def __init__(self,name,age,Father_name, Mother_name):
        Father.__init__(self,Father_name)
        Mother.__init__(self,Mother_name)
        self.name=name
        self.age=age
    def parent_work(self):
        print(super().work1())
        print(super().work2())
s = son("siva", 32,"pavan","satya")
print(s.Father_name)
print(s.Mother_name)
#print(s.work())
#print(s.work())
print(s.parent_work())
```
