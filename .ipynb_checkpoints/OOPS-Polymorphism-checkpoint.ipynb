- `POLYMORPHISM`: In python `poly` means many and `morphism` means forms(behaviours)
- In polymorphism single function shows different behaviours based on input context.
- In python polymorphism can obtained thriugh the following ways:
     1. Through operators
     2. polymorphism Through Functions
     3. polymorphism Through methods
Types in polymorphism Through methods:
-  Runtime Polymorphism(`method overriding`)
-  compiletime polymorphism (`method overloading`)

1. `POLYMORPHISM THROUGH OPERATORS`: Here Single operator performs shows different actions based on input context.

   Eg: `+` operator: It performs addition operation when inputs are integers also It performs string concatination when inputs are on strings.

   eg for addition operator
   3+5 = 8 <br>
   code + gnan = codegnan
   eg for multiplication operator

   
   2 * 3 = 6<br>
   2 * code = "code" "code"  #repetation

2. `POLYMORPHISM THROUGH FUNCTIONS`: Here a function shows different action based on input context.

   eg: len("hello")->returns string length

   
      len([1,2,3,4])->returns list length

3. `POLYMORPHISM THROUGH METHODS`:

    - `Method Overriding` : A method implemented in child class are defined in child class that is already exist in parent           class with the same name.
    -  Whenever we call that method through the child class object, the child class method overrides the parent class      `         method,This is known as method overriding
    - `Method Overloading` : A class contains multiple methods within the same name with different parameters list,In that           case last defined method overrids the previous defined method.This is called method overloading
    -  Python do not directly supports method overloading but we can achieve the same functionality by using `default                arguments,variable length positional or variable length keyword arguments`



```python
#method overriding
class Person:   #parent class
    def __init__(self,name,age,gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):  #child classs
        return f'My name is {self.name} and I am {self.age} years old'
        
class Student(Person):
    def __init__(self,name,age,gender,branch,stdid):
        super().__init__(name,age,gender)
        self.branch = branch
        self.stdid = stdid
    def details(self):
        return f'My name is {self.name} and i am from {self.branch} branch'
        
s = Student("rishi",22,"female","ece",101)
print(s.name)
print(s.details())
```

```python
#method overriding
#even if we define different number of parameters in methods,always child method overrides parent class method
class Person:   #parent class
    def __init__(self,name,age,gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):  #child classs
        return f'My name is {self.name} and I am {self.age} years old'
        
class Student(Person):
    def __init__(self,name,age,gender,branch,stdid):
        super().__init__(name,age,gender)
        self.branch = branch
        self.stdid = stdid
    def details(self):
        print(super().details())    #calling the parent class
        return f'My name is {self.name} and i am from {self.branch} branch'
        
s = Student("rishi",22,"female","ece",101)
print(s.name)
print(s.details())
```

```python
#MethodOverloading: 
class Math:
    def add(self,a,b):
        return a+b
    def add(self,a,b,c):
        return a+b+c
    def add(self,a,b,c,d):
        return a+b+c+d
m = Math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
```

```python
#as python cannot support method overloadibg directly we are using the default or variable length positional or variable length keyword arguments(i.e arbitary arguments)
class Math:
    def add(self,a,b,c=0,d=0):
        return a+b+c+d
m = Math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
```

```python
##Magic methods or special methods
#aoperators
#__str__()
class Point:
    def __init__(self,x):
        self.x = x
    def __str__(self):  #String Magic method
        return f"This is Point class and Point value is {self.x}"
    def __add__(self,other):
        return Point(self.x+other.x)
p1 = Point(5)
print(p1)
p2 = Point(3)
#print(p1.x+p2.x)
p3 = p1+p2
print(type(p3))
print(p3)
```

```python
##Magic methods or special methods
#__str__()
class Point:
    def __init__(self,x):
        self.x = x
    def __str__(self):  #String Magic method
        return f"This is Point class and Point value is {self.x}"
    def __mul__(self,other):
        return(self.x*other.x)
    def __sub__(self,other):
        return(self.x-other.x)
    def __gt__(self,other):
        return(self.x>other.x)
    def __lshift__(self,other):
        return(self.x<<other.x)
    def __lt__(self,other):
        return(self.x<other.x)
    def __or__(self,other):
        return(self.x|other.x)
    def __mod__(self,other):
        return(self.x % other.x)
    def __floordiv__(self,other):
        return(self.x//other.x)
        
p1 = Point(5)
p2 = Point(3)
print(p1*p2)
print(p1-p2)
print(p1>p2)
print(p1<p2)
print(p1<<p2)
print(p1<p2)
print(p1|p2)
print(p1%p2)
print(p1//p2)
```
