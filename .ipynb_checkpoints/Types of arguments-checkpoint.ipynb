```python
#Types of arguments
#1.Positional arguments: Here arguments are assigned to parameters as per the sequential order 
#function defination block
def sub(x:int,y:int)->int:
    z = x-y
    return z
#function defination block
a,b = 20,10
print(sub(a,b))
print(sub(b,a))
```

```python
#keyword Argumengts: arguments are explicitly passed with parameter names
#function defination block
def sub(x:int,y:int)->int:
    z = x-y
    return z
#function defination block
a,b = 20,10
print(sub(x = a,y = b))
print(sub(y = b,x = a))
```

```python
def add(x:int,y:int)->int:
    z = x+y
    return z
#function defination block
a,b = 20,10
print(add(x = a,y = b))
print(add(y = b,x = a))
print(add(10,y=5))  #frst positional argumenent should be passed then the key word srgument
```

```python
#3 number addition
def add(x:int,y:int,z:int)->int:
    a = x+y+z
    return a
#function defination block
a,b,c = list(map(int,input().split()))
print(add(a,b,c))
```

```python
#4 number addition
def add(w:int,x:int,y:int,z:int)->int:
    a = w+x+y+z
    return a
#function defination block
a,b,c,d = list(map(int,input().split()))
print(add(a,b,c,d))
```

```python
#Default arguments: The default arguments are optional arguments, if you dont pass value for default arguments,in that case it assigns default value otherwise,it assigns new argument values
#4 number addition
def add(w:int,x:int,y:int=0,z:int=0)->int:
    a = w+x+y+z
    return a
#function defination block
a,b,c,d = list(map(int,input().split()))
print(add(a,b))
print(add(a,b,c))
print(add(a,b,c,d))
```

```python
#4 number multiplication
def mul(w:int,x:int,y:int=1,z:int=1)->int:
    a = w*x*y*z
    return a
#function defination block
a,b,c,d = list(map(int,input().split()))
print(mul(a,b))
print(mul(a,b,c))
print(mul(a,b,c,d))
```

```python
#variable length positional arguments
def add(x,y=0,*args):
    print(type(args))
    print(args)
    s = 0
    for val in args:
        s+= val
    return s
print(add(10,20,1,2,3,4,5))    
```

```python
#variable length keyword arguments
def add(x,y=0,*args,**kwargs):
    print(type(kwargs))
    print(args)
    print(kwargs)
    print(kwargs.values())
    print(kwargs.keys())

print(add(1,2,3,4,5 , name="rishi",age=21,course="PFS"))
```

```python
def fun(x):
    print(id(x))
a = 10
print(id(a))
fun(a)
```

```python
def fun(x):
    print(id(x))
    x+=2
    print('after add 2')
    print(id(x))
a = 10
print(id(a))
fun(a)
```

```python
#In python every type function calling block shares the arguments reference to parameters
#  if the arguments are immutable data types it ,then behave like pass by value and if arguments are mutable datatypes(i.e list,set,dict) then behaves like
#      pass by reference
## Pass by Value:
def fun(x):
    print(id(x))
    x+=2
    print('after add 2')
    print(id(x))
    print(x)
a = 10
fun(a)
print(id(a))

print(a)
```

```python
# Pass by Reference
def add_ele(lst,ele):
    lst.append(ele)

arr = [1,2,3,4]
print("Before Function execution arr value:",arr)
add_ele(arr,100)
print("After Function execution arr value:",arr)
```

```python
# Pass by Reference
def add_ele(lst,ele):
    lst = [1,2,3,4]

arr = [1,2,3,4]
print("Before Function execution arr value:",arr)
add_ele(arr,100)
print("After Function execution arr value:",arr)
```
