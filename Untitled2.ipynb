```python
from datetime import datetime
current_timestamp = datetime.now() #to get the current timestamp
print(current_timestamp)
today = current_timestamp.date() #to extract date from current timestamp
print(today)
cur_time = current_timestamp.time() #to extract time from current timestamp
print(cur_time)
#extract day,month, and year
day = today.day
month = today.month
year = today.year
print(day,month,year)
print(cur_time.hour,cur_time.minute,cur_time.second)
#changing date format
print(today.strftime("%d / %m / %Y")) #Y-4 DIGIT YEAR
print(today.strftime("%d / %m / %y")) #y-2 DIGIT YEAR
print(today.strftime("%d / %B / %Y")) #B-full month name
print(today.strftime("%d / %b / %Y")) #b-abbreviated month name
print(today.strftime("%d / %A / %Y")) #A-fullweek day
print(today.strftime("%d / %a / %Y")) #a-abbreviated weekday name

#changing time format
print(cur_time.strftime("%H:%M:%S")) #24 hrs format
print(cur_time.strftime("%I:%M%p")) #12 hrs format
```

```python
#Timer implementation
import time
from IPython.display import clear_output
hour_counter = minute_counter = second_counter = 0
while True:
    print(f"{hour_counter} : {minute_counter} : {second_counter}")
    time.sleep(1) #It pauses the execution for 1 second
    second_counter += 1
    if second_counter == 60:
        minute_counter += 1
        second_counter = 0 #reset second counter to 0
        if minute_counter == 60:
            hour_counter += 1
            minute_counter = 0 #reset minute counter
            if hour_counter == 24:
                hour_counter = 0
    clear_output(wait = True)
            
```
