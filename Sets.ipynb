- ## Set characteristics

```python
# set unordered collection
# set returns unique values and do not allow diplicates
s = {1,2,3,2,4,"hi",'a',4.3,(2,0,3)}
print(s)
# print(s[0]) #set do not support indexing
```

```python
#list cant be used as set elements
s = {[1,2,3],'hi',3,7}
print(s)
```

```python
# empty set representation
s1 = {} #this is already used to represent empty set of dictionary
s2 = set()
print(type(s1),type(s2))
```

- ### Set Methods

```python
# Element adding method
# add(ele): It adds the element to set if element not in set,otherwise nothing
s = {1,2,4,2,3,6,4,5,5,6}
print(s)
s.add(100)
s.add(5)
s.add('code')
print(s)
```

```python
# Element removing methods
# remove(item): It removes the item from set if it exists in set otherwise, 'keyerror'
# discard(item): It removes the item from set if it exists in set otherwise nothing
s = {1,2,3,4,5}
s.remove(5)
print(s)
#s.remove(100)  #keyerror
s.discard(3)
print(s)
s.discard(100)
print(s)
```

- ## Sets operations
- union(set): It returns all the unique elements from all sets
  

```python
s1 = {1,3,5,7}
s2 = {4,5,6,7}
s3 = {1,2,8,9}
union = s1.union(s2,s3)
print(union)
res = s1|s2|s3
print(res)
```

```python
#Intersection(set) :it returns common elements
s1 = {1,3,5,7}
s2 = {4,5,6,7}
s3 = {1,2,8,9}
intersection = s1.intersection(s3)
print(intersection)
intersection = s1.intersection(s2,s3)
print(intersection)
res = s1&s2&s3
print(res)
```

```python
# difference(set):It returns the set elements those are not in other sets
s1 = {1,3,5,7}
s2 = {4,5,6,7}
s3 = {1,2,8,9}
difference = s1.difference(s3)
print(difference)
difference = s1.difference(s2,s3)
print(difference)
res = s1-s2-s3
print(res)
print(s1&s2-s3)
print(s1|s2-s3)
```

```python
# sets operations
# symmetric_difference(set): it returns non common elements from all sets
s1 = {1,3,5,7}
s2 = {4,5,6,7}
s3 = {5,1,2,8,9}
sy_diff = s1.symmetric_difference(s2)
print(sy_diff)
res = s1^s2^s3
print(res)
```

```python
# issubset(set)
# issuperset(set)
# isdisjoint(set)
s1 = {1,3,5,7}
s2 = {1,3,5,7}
s3 = {5,1,2,8,9}
subset = s1.issubset(s2)
print(subset)
superset = s2.issuperset(s3)
print(superset)
disjoint = s1.isdisjoint(s3)
print(disjoint)
s = s1,s2,s3
print(dir(s))
```
