```python
import re
```

```python
#FUNCTIONS 
#1.match(pat,string)...(true case)
string = "The sky is blue and ocean is blue"
res = re.match("The",string)
print(res)
print(res.span()) #it returns the matched pattern positions
print(res.group()) #it returns the matched pattern result
```

```python
#FUNCTIONS 
#1.match(pat,string): it checks weather the pattern starts with pattern or not
#false case
try:
    string = "The sky is blue and ocean is blue"
    res = re.match("sky",string)
    print(res)
    print(res.span()) #it returns the matched pattern positions
    print(res.group()) #it returns the matched pattern result
except Exception as e:
    print("There is no matched pattern")
```

```python
#2.search(pat,string)...(true case):it checks the entire string and retirns the first matched occurance
string = "The sky is blue and ocean is blue"
res = re.search("blue",string)
print(res)
print(res.span()) #it returns the matched pattern positions
print(res.group())#it returns the matched pattern result
```

```python
#search(pat,string)....(false case)
try:
    string = "The sky is blue and ocean is blue"
    res = re.search("z",string)
    print(res)
    print(res.span()) #it returns the matched pattern positions
    print(res.group()) #it returns the matched pattern result
except Exception as e:
    print("There is no matched pattern")
```

```python
#findall(pat,string):it returns all matched patterns as list
string = "The sky is blue and ocean is blue"
res = re.findall("blue",string)
print(res)
```

```python
#findall(pat,string)
string = "The sky is blue and ocean is blue"
res = re.findall("[abc]",string)
print(res)
```

```python
#findall(false case)
try:
    string = "The sky is blue and ocean is blue"
    res = re.findall("z",string)
    print(res)
except Exception as e:
    print("There is no matched pattern")
```

```python
#re.split(pat,string):it splits the string wherever the pattern is matched 
string = "The sky is blue and ocean is blue"
res = re.split("[ae]",string)
print(res)
```

```python
#re.split
try:
    string = "The sky is blue and ocean is blue"
    res = re.split("[obe]",string)  #[obe] splits individually 'o','b','e'
    print(res)
except:
    print("There is no matched pattern")
```

```python
#sub(old_pattern,new_pat,string):it replaces or substitutes the old_pat with new_pat
try:
    string = "The sky is blue and ocean is blue"
    res = re.sub("blue","red",string)  
    print(res)
except:
    print("There is no matched pattern")
```

```python
##META CHARACTERS
try:
    string = "The sky is blue and ocean is blue"
    res = re.findall("[isabce]+",string)  #META CHARACTERS
    print(res)
except Exception as e:
    print("There is no matched pattern")
```

```python
##META CHARACTERS
try:
    string = "The sky is blue and ocean is blue"
    res = re.findall("[a-zA-Z]+",string)  #META CHARACTERS
    print(res)
except Exception as e:
    print("There is no matched pattern")
```

```python
##META CHARACTERS:when integers are mentiones
try:
    string = "The sky is blue and ocean is bl6ue 4557"
    res = re.findall("[a-zA-Z]+",string)  #META CHARACTERS
    print(res)
except Exception as e:
    print("There is no matched pattern")
```

```python
##META CHARACTERS:when integers are mentiones
#when we want to return only numbers
try:
    string = "The sky is blue and ocean is bl6ue 4857"
    res = re.findall("[0-9]",string)  #META CHARACTERS  +not included
    print(res)
except Exception as e:
    print("There is no matched pattern")
```

```python
#validate mobile number using meta characters
try:
    mobile = input()
    res = re.fullmatch("^[6-9][0-9]{9}", mobile)
    print(res)
    if res:
        print(f'{mobile} is valid')
    else:
        print(f'{mobile} is invalid')
except:
    print("There is no matched pattern")
```

```python
password = input("Enter Password: ")
if (len(password) >= 8 and
    re.search(r"[A-Z]", password) and
    re.search(r"[a-z]", password) and
    re.search(r"\d", password) and
    re.search(r"[!@#$%^&*]", password)):
    print("Strong Password")
else:
    print("Weak Password")
```
