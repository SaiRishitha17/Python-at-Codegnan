- ### Input formating

```python
# Reading integer from user
num = int(input('Enter a number:'))
print(type(num))
print(num)
```

```python
# Reading float value from user
val = float(input())
print(type(val))
print(val)
```

```python
# complex reading from user
val = complex(int(input()),int(input())) #input() are for twice because we are passing 2 values here
print(type(val))
print(val)
```

```python
# boolean reading from user
val = bool(int(input()))
print(type(val))
print(val)
```

```python
#reading string values from user
s = input()
print(type(s))
print(s)
```

```python
# reading list of integers from user
lst = input()
print(lst)

```

```python
lst = input().split()
print(lst)
```

```python
lst = map(int,input().split())
print(lst)
```

```python
lst = list(map(int,input().split()))
print(lst)
```

```python
#reading list of strings
lst = list(map(str,input().split()))
print(lst)
```

```python
# reading list of strings
lst = list(map(str,input().split(',')))
print(lst)
```

```python
#approach 2
lst = input().split()
print(lst)
```

```python
#reading tuple of integers from user
lst = tuple(map(int,input().split()))
print(lst)
```

```python
#reading set of integers from users
lst = set(map(int,input().split()))
print(lst)
```

```python
# eval(): This function evaluates the string type expression
val = eval(input())
print(type(val))
print(val)
```

```python
print('3+5')
print(eval('3+5'))
```

- ### Output formatting

```python
# 1. string formatting using comma  
name = "rishi"
age = "20"
print("My name is"  , name , "and i am"  , age  , "years old")
```

```python
# using string concatinatiom  #old style formatting
name = "rishi"
age = "20"
print("My name is"  + name + "and i am"  +  age  + "years old")
```

```python
# 2.Access specifiers
# string - %s
# integer - %d
# float - %f
name = "rishi"
age = 20
print("My name is %s and i am %d years old"%(name,age))
```

```python
# 2.Access specifiers
# string - %s
# integer - %d
# float - %f
name = "rishi"
age = 20
print("My name is %s and i am %d years old"%(age,name))
```

```python
# 3.dot format
name = "rishi"
age = 20
print("My name is {} and i am {} years old".format(name,age))
print("My name is {} and i am {} years old".format(age,name))
```

```python
# 4.f-format
name = "rishi"
age = 20
print(f"My name is {name} and i am {age} years old")
print(f"My name is {age} and i am {name} years old")
```
