```python
# integer to float,bool,string
num = 10
print(float(num))    #since float,bool and string are builtin functions...
print(bool(num))
print(str(num))
print(complex(num))
```

```python
# float to int,bool and string
val = 10.256
num = int(val)
print(type(num),num)
print(bool(val))
str_val = str(val)
print(type(str_val),str_val)
print(str_val[:4])
```

```python
val = 102.43256
strval = str(val)
ind = strval.find('.')
print(strval[:ind])
```

```python
# bool to integer,float and string
b = True
print(int(b))
print(float(b))
print(str(b))
```

```python
#string to int,float,bool,list,tuple,set
#if the string is a number we even have float
s = "123456"
print(int(s))
print(float(s))
print(bool(s))
print(list(s))
print(tuple(s))
print(set(s))
```

```python
#string to int,float,bool,list,tuple,set
#if string value is float
s = "123456.1205"
print(int(float(s)))
print(float(s))
print(bool(s))
print(list(s))
print(tuple(s))
print(set(s))
```

```python
#string to int,float,bool,list,tuple,set
#if string value is float
s = "123456.1205"
print(int(float(s)))
print(float(s))
print(bool(s))
print(list(s))
print(tuple(s))
print(set(s))
```

```python
# list to string,tuple,set
s = [12,3,45,2.34,'rishi',23]
print(str(s))
print(tuple(s))
print(set(s))
```

```python
# tuple to string,list,set
t = (1,3,4.4,'hello')
print(str(t))
print(list(t))
print(set(t))
```

```python
# set to string,list,tuple
s = {2,3,45,11,23,55,'hello',10.5}
print(str(s))
print(list(s))
print(tuple(s))
```

```python
#dictionary to string,list,tuple,set
s = {"name":"rishi","batch":57, "course":"PFS","age":21}
print(str(s))
print(list(s))
print(tuple(s))
print(set(s))
```
