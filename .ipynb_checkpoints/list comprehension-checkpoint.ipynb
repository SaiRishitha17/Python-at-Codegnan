```python
# generate 1-n numbers using list comprehension
n = int(input())
res = [x for x in range(1,n+1)]
print(res)
```

```python
# generate 1-n numbers using list comprehension
n = int(input())
res = [x*2 for x in range(0,n)]
print(res)
```

```python
#find the first n cubic numbers
n = int(input())
res = [x**3 for x in range(1,n+1)]
print(res)
```

```python
#finding add numbers between n1 and n2
n1,n2 = map(int,input().split())
res = [x for x in range(n1,n2+1) if x%2==1]
print(res)
```

```python
# in a given arr of integers perform following operations
#if even - find square of a number
#if odd - find squareroot value(upto 2 decimal places)
arr = list(map(int,input().split()))
res = [x**2 if x%2==0 else round(x**0.5,2) for x in arr]
print(res)

    
```

```python
#b2 dices combination values
res = [(x,y) for x in range(1,7) for y in range (1,7)]
print(res)
```

```python
#return the duce combinations if sum of two face value is divisible by 3
res = [(x,y) for x in range(1,7) for y in range (1,7) if (x+y)%3==0]
print(res)
```

```python
res = [(x,y) for x in range(1,7) for y in range (1,7) if x==4 or y==4]
print(res)
```

```python
# count the number of vowels existing in the string
s = input().lower()
res = [x for x in s if x in "aeiou"]
print(res)
print(len(res))
```

- ### Set Comprehension

```python
#1 to n numbers
n = int(input())
res = {x for x in range(1,n+1)}
print(type(res))
print(res)
```

```python
#1 to n numbers
n = int(input())
res = {x:x**2 for x in range(1,n+1)}
print(type(res))
print(res)
```

- ### Tuple comprehension

```python

#1 to n numbers
n = int(input())
res = tuple(x for x in range(1,n+1))  #tuple is not directly generated so we are using a keyword "Tuple" 
print(type(res))
print(res)
```

```python
# assert keyword : This keyword is used for testing
s = int(input())
if s>=18:
    print('eligible to vote')
else:
    print('not eligible')
```

```python
## reverse a string without using slicing
string = input()
rev = ''
for char in string:
    rev = char + rev
print(rev)
```

```python
n = int(input())
res = ""
while n>0:
    rem = n%2
    res = str(rem) + res
    n = n // 2
print(res)
rev = ""
for char in res:
    rev = char + rev
print(rev)
print("Pallindrome" if res == rev else "Not Pallindrome")
```
