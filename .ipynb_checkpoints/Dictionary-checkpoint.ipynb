- ## Dictionary characteristics {}

```python
# Dictionary ordered collection
details = {"Name": "Rishi", "course":"PFS", "Batch":57, "age":20}
print(details)
```

```python
# Dictionary values access by using keys
# updating dictionary values
# adding new items to dictionary
details = {"Name": "Rishi", "course":"PFS", "Batch":57, "age":20}
print(details["course"])
# print(details['location']) #keyerror: 'location'
# updating dictionary values
details['age'] = 21
print(details['age'])
# adding new items to dictionary
details['location'] = "Hyderabad"
print(details)
```

```python
details = {"age":25, "Name": "Rishi", "course":"PFS", "Batch":57, "age":20}
print(details)
```

```python
# list can't be used as dictionary key
# dictionary keys are unique
details = {"age":25, "Name": "Rishi", "course":"PFS", "Batch":57, "age":20 ,"marks":[12,23,22]}
print(details["marks"])
```

- ### Dictionary Methods
- get(key, default_value): It returns the value if key is existing in dictionary, otherwise it returns default value

```python
# dictionary keys are unique
details = {'Name': 'Rishi', 'course': 'PFS', 'Batch': 57, 'age': 21, 'location': 'Hyderabad'}
print(details['Name'])
#print(details['marks']) #results key error as the key is not defined,inorder to avoid the key error we use .get and we have no change in the output by using .get
print(details.get('Name'))
print(details.get('marks'))
print(details.get('marks',"marks not available"))  #it returns dafault values because,

```

```python
##update(new_dictionary) : This method updated the value if new_dictionary keys already exist in Dictionary,otherwise it adds new items to dictionary
```

```python
details = {"age":25,'Name': 'Rishi', 'course': 'PFS', 'Batch': 57,}
details.update({"Batch":58, "course":"java", "location": ":hyd"})    # here existing keys will be updating and non existing will be added to the items
print(details)
```

- keys() :This method returns the list dictionary keys
- values() :This method returns the list dictionary values
- items() :This methods returns the list of Tuple of key-value pairs

```python
details = {'age': 25, 'Name': 'Rishi', 'course': 'java', 'Batch': 58, 'location': ':hyd'}
keys = details.keys()
print(details.keys())
print(type(keys))
print(keys)
print(details.values())
print(details.items())

```

- popitem(): It return and removes the last item as tuple
- pop(key): This method removes the specified key item and return the specifies key's value,otherwise key error

```python
details = {'age': 25, 'Name': 'Rishi', 'course': 'java', 'Batch': 58, 'location': ':hyd'}
item = details.popitem()
print(item)
print(details)
val = details.pop('Name')
print(val)
print(details)
```

```python
# Membership
details = {'age': 25, 'Name': 'Rishi', 'course': 'java', 'Batch': 58, 'location': ':hyd'}
print('rishi' in details)
print(25 in details)
print('course' in details)
```
