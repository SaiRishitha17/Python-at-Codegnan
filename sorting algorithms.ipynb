```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
numbers = list(map(int, input("Enter the elements: ").split()))
print("\nOriginal List:")
print(numbers)
bubble_sort(numbers)
print("\nSorted List:")
print(numbers)
```

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_index = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
numbers = list(map(int, input("Enter the elements: ").split()))
print("\nOriginal List:")
print(numbers)
selection_sort(numbers)
print("\nSorted List:")
print(numbers)
```

```python
# Insertion Sort in Python
numbers = [45, 12, 89, 23, 7, 34]
print("Original List:", numbers)
for i in range(1, len(numbers)):
    key = numbers[i]
    j = i - 1

    while j >= 0 and numbers[j] > key:
        numbers[j + 1] = numbers[j]
        j -= 1

    numbers[j + 1] = key

print("Sorted List:", numbers)
```

```python
# Student Marks Organizer using Insertion Sort

students = [ ["Riya", 78],["Aman", 56],["Neha", 92],["Kiran", 65],["Rahul", 81]]
print("Student Marks Before Sorting")
for student in students:
    print(f"{student[0]} - {student[1]}")
# Insertion Sort based on marks
for i in range(1, len(students)):
    key = students[i]
    j = i - 1
    while j >= 0 and students[j][1] > key[1]:
        students[j + 1] = students[j]
        j -= 1
    students[j + 1] = key
print("\nStudent Marks After Sorting")
for student in students:
    print(f"{student[0]} - {student[1]}")
```

```python
def merge(left, right):
    result = []
    i = 0
    j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)
numbers = [45, 12, 78, 34, 23, 9, 56]
print("Original List:", numbers)
sorted_list = merge_sort(numbers)
print("Sorted List:", sorted_list)
```

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = []
    middle = []
    right = []
    for num in arr:
        if num < pivot:
            left.append(num)
        elif num == pivot:
            middle.append(num)
        else:
            right.append(num)
    return quick_sort(left) + middle + quick_sort(right)
numbers = []
n = int(input("Enter the number of elements: "))
for i in range(n):
    num = int(input(f"Enter element {i + 1}: "))
    numbers.append(num)
print("\nOriginal List:", numbers)
sorted_numbers = quick_sort(numbers)
print("Sorted List:", sorted_numbers)
```
