```python
#vowels and consonents
s = input().lower()
c = 0
v = 0
for char in s:
    if char in "aeiou":
        v+=1
    else:
        c+=1
print(f'vowels {v},consonents {c}')
    
```

- #### while loop

```python
#print 1 to n  numbers using while loop
n = int(input())
i = 1  #initialization
while i<=n:
    print(i)
    i+=1
```

```python
#print 1 to n  numbers using while loop(reverse)
n = int(input())
i = 1  #initialization
while i<=n:
    print(n)
    n-=1
```

```python
#prime numbers
n = int(input())
factors = list()
for i in range(1,n+1):
    if n%i == 0:
        factors.append(i)
if len(factors)==2:
    print(f'{n} is prime')
else:
    print(f'{n} is not prime')




```

```python
#prime number
n = int(input())
for i in range(2,n//2+1):
    if n%i==0:
        print(f'{n}is not a prime number')
        break
else:
    print(f'{n} is prime number')
```

```python
#prime number
m = int(input())
for n in range (2,m):
    for i in range(2,n//2+1):
        if n%i==0:
            print(f'{n}is not a prime number')
            break
else:
    print(f'{n} is prime number')
```
