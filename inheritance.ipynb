```python
##Multilevel Inheritance
class Product:
    def __init__(self, name):
        self.name = name

    def product_info(self):
        return f"Product: {self.name}"
class Electronics(Product):
    def __init__(self, name, brand):
        super().__init__(name)
        self.brand = brand
class Mobile(Electronics):
    def __init__(self, name, brand, price):
        super().__init__(name, brand)
        self.price = price

    def display(self):
        return f"{self.product_info()}, Brand: {self.brand}, Price: ₹{self.price}"
m = Mobile("Galaxy S25", "Samsung", 75000)
print(m.display())
```

```python
##Multiple Inheritance
class Payment:
    def payment_status(self):
        return "Payment Successful"
class Delivery:
    def delivery_status(self):
        return "Order Delivered"
class Order(Payment, Delivery):
    def __init__(self, order_id):
        self.order_id = order_id
o = Order(101)

print("Order ID:", o.order_id)
print(o.payment_status())
print(o.delivery_status())
```

```python
##Hierarchical Inheritance
class User:
    def __init__(self, name):
        self.name = name

class Customer(User):
    def shop(self):
        return f"{self.name} placed an order."

class Seller(User):
    def sell(self):
        return f"{self.name} listed a new product."


c = Customer("Rishitha")
s = Seller("Ramu")

print(c.shop())
print(s.sell())
```

```python
#hybrid inheritance
class User:
    def __init__(self, name):
        self.name = name
class Customer(User):
    def place_order(self):
        return f"{self.name} placed an order."
class PrimeMember(User):
    def free_delivery(self):
        return "Eligible for free delivery."
class PremiumCustomer(Customer, PrimeMember):
    def discount(self):
        return f"{self.name} gets an extra 20% discount."
p = PremiumCustomer("Rishitha")

print(p.place_order())
print(p.free_delivery())
print(p.discount())
```
