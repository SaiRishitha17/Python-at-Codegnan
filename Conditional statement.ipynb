```python
#generate total bill if total bill is more than 500 then give 10% discount otherwise 0%
total_bill = int(input())
if total_bill>500:
    discount = total_bill * 0.1
    print(f'Total bill After 10% discount: {total_bill - discount}')
else:
    print(f'Total bill After 0% discount:{total_bill}')
```

```python
#check weather the product stock is available or not(read product name and stock)
product = input()
stock = int(input())
if stock > 0:
    print(f"currently {product}:{stock} items available")
else:
    print(f"currently {product} out of stock")
```

```python
#in a shop contains some items along with prices like {'pen':5,'book':20,'pencil':3}

products = {'pen':5,'book':20,'pencil':3}
item = input().lower()
quantity = int(input())
if item in products:
    total_amount = products.get(item) * quantity
    print(f'Total amount of {quantity} {item} is: {total_amount}')
else:
    print(f'{item} not available')
```

```python
# divisable by 3 - Fizz
# divisable by 5 - Buzz
# both - FizzBuzz
num = int(input())
if num%3==0:
    print(f'{num} is Fizz')
elif num%5==0:
    print(f'{num} is Buzz')
elif num%3==0 and num%5==0:
    print(f'{num} is FizzBuzz')
else:
    print(f'{num} is invalid') #here when we give 15 as input the output should be FizzBuzz,but we get fizz as the frst condition is true the it goes out of code.so we need to give priority wise
```

- ### Nested if

```python
# check weather the number is +ve even,-ve even,+ve odd,-ve odd
n = int(input())
if n>=0:
    if n%2==0:
        print(f'n is positive even')
    else:
        print(f'n is positive odd')
else:
        if n%2==0:
            print(f'n is negative evn')
        else:
            print(f'n is negative odd')
        
    
    
```

```python
n1,n2,n3 = map(int,input().split())
if n1>n2:
    if n1>n3:
        print(f'n1')
    else:
        print(f'n3')
else:
        if n2>n3:
            print(f'n2')
        else:
            print(f'n3')
            
```
