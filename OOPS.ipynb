```python
class Mobile:
    company = "Samsung"   # class attributes
    def __init__(self,model,color,price): #initialisation constructor(magic method)
## any methods starts and ends with dunderscore those methods are called as magic methods
        self.model = model  #Instance Attributes
        self.color = color
        self.price = price
        #Instance Attributes
    def details(self):
        return f"{self.company} {self.model} is available in {self.color} and costs {self.price}"

s1 = Mobile("Galaxy A35","Ice Blue",28999)
s2 = Mobile("Galaxy S24", "Onyx Black",64999)
s3 = Mobile("Galaxy F55","Apricot Crush",26999)
print(s1.model)
print(s2.details())
print(s1.details())
print(s3.details())
```
