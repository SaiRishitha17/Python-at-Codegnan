- ### Scope of variables

```python
#1.Local scop
#2.Enclosing scope
#3.Global scope
#4.Built-in scope
x=10  #global scope variable
def outer():
    y=20 #enclosing scope variable
    def inner():
        z=30 #local scope variable
        print(z,y,x) #built-in scope variable
    inner()
outer()
```

```python
#using Global scope variable in local scope
x = 100
def outer():
    y = x+1
    print("Local scope variable value",y)
outer()
print("Global scope variable value:",x)
```

```python
#updating the Global scope variable in local scope
#cannot update global variable in global scope,directly
x = 100
def outer():
    x = x+1  #cannot update the values,we can only access...
    print("Local scope variable value",x)
outer()
print("Global scope variable value:",x)

```

```python
#updating the Global scope variable in local scope by using 'global' keyword
x = 100
def outer():
    global x
    x = x+1  #cannot update the values,we can only access...
    print("Local scope variable value",x)
outer()
print("Global scope variable value:",x)

```

```python
# updating enclosing scope variable value in local scope
def outer():
    y=20 #enclosing scope variable
    def inner():
        y=y+10 #local scope variable
        print(y)
    inner()
outer()
```

```python
# updating enclosing scope variable value in local scope by using keyword 'nonlocal'
def outer():
    y=20 #enclosing scope variable
    def inner():
        nonlocal y
        y=y+10 #local scope variable
    inner()
    print("Enclosing scope variable value:",y)
outer()
```

```python
x=100
def outer():
    x=200
    x=x+2
    print('local scope variable x value:',x)
outer()
print('Global scope variable x value:',x)
```
