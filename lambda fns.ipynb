```python
from functools import reduce
students = [
    {"name": "Rishi", "marks": 92},
    {"name": "Rohith", "marks": 76},
    {"name": "divya", "marks": 88},
    {"name": "priya", "marks": 65},
    {"name": "rahul", "marks": 95}]
# Extract student names
names = list(map(lambda student: student["name"], students))
# Find students who scored 80 or above
top_students = list(filter(lambda student: student["marks"] >= 80, students))
# Calculate total marks
total_marks = reduce(lambda total, student: total + student["marks"], students, 0)
# Calculate average marks
average = total_marks / len(students)
print("Student Names:")
print(names)
print("\nStudents Scoring 80 or Above:")
for student in top_students:
    print(student)
print("\nTotal Marks:", total_marks)
print("Average Marks:", round(average, 2))
```
