```python
from abc import ABC, abstractmethod
class Mobile(ABC):

    @abstractmethod
    def ram(self):
        pass
    @abstractmethod
    def rom(self):
        pass
    @abstractmethod
    def battery(self):
        pass

#concrete method
    def display(self, Hzs):
        return f"This Mobile Display Quality is {Hzs} Hzs"
class Oppo(Mobile):
    def __init__(self, model, battery_size, display_quality):
        self.model = model
        self.battery_size = battery_size
        self.display_quality = display_quality
    def ram(self, ram_size):
        return f"The Oppo Mobile {self.model} has {ram_size}GB RAM"
    def rom(self, rom_size):
        return f"The Oppo Mobile {self.model} has {rom_size}GB ROM"
    def battery(self):
        return f"The Oppo Mobile {self.model} has {self.battery_size} mah Battery capacity"
    def camera(self, primary, secondary):
        return f" The Oppo {self.model} mobile have primary camera with {primary}MP and secondary camera with {secondary}MP"
 # m = Mobile()- TypeError: Can't instantiate abstract class Mobile without an implementation for abstract methods 'battery', 'ram', 'rom'
reno15 = Oppo("reno15", 5000, 120)
print(reno15.ram(4))
print(reno15.rom(128))
print(reno15.battery())
print(reno15.camera(32,52))
print(reno15.display(120))
```

```python
from abc import ABC, abstractmethod
import math
# Abstract Class
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
    @abstractmethod
    def details(self):
        pass
# 2D Shape - Circle
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius ** 2
    def perimeter(self):
        return 2 * math.pi * self.radius
    def details(self):
        print("2D Shape: Circle")
        print(f"Radius: {self.radius}")
        print(f"Area: {self.area():.2f}")
        print(f"Perimeter: {self.perimeter():.2f}")
# 2D Shape - Rectangle
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
    def perimeter(self):
        return 2 * (self.length + self.width)
    def details(self):
        print("\n2D Shape: Rectangle")
        print(f"Length: {self.length}")
        print(f"Width: {self.width}")
        print(f"Area: {self.area()}")
        print(f"Perimeter: {self.perimeter()}")
# 3D Shape - Cube
class Cube(Shape):
    def __init__(self, side):
        self.side = side
    def area(self):
        return 6 * self.side ** 2
    def volume(self):
        return self.side ** 3
    def details(self):
        print("\n3D Shape: Cube")
        print(f"Side: {self.side}")
        print(f"Surface Area: {self.area()}")
        print(f"Volume: {self.volume()}")
# 3D Shape - Sphere
class Sphere(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 4 * math.pi * self.radius ** 2
    def volume(self):
        return (4/3) * math.pi * self.radius ** 3
    def details(self):
        print("\n3D Shape: Sphere")
        print(f"Radius: {self.radius}")
        print(f"Surface Area: {self.area():.2f}")
        print(f"Volume: {self.volume():.2f}")
# Objects
circle = Circle(7)
rectangle = Rectangle(10, 5)
cube = Cube(4)
sphere = Sphere(3)
circle.details()
rectangle.details()
cube.details()
sphere.details()
```
